In Kurztest 3 (Aufgabe 2) wird eine Stammfunktion $F(x)$ gesucht, bei der $F(1) = 2$ gilt. In den Moodle-Aufgaben kommen Rotationskörper vor.

```python
import sympy as sp

x, C = sp.symbols('x C')

# ==========================================
# 1. HIER ANPASSEN FÜR ANFANGSWERTPROBLEM:
f_x = 3*x**2 - 4*x + 1
x_wert = 1
F_wert = 2
# ==========================================

print("--- ANFANGSWERTPROBLEM (Stammfunktion mit C) ---")
# 1. Unbestimmtes Integral berechnen
stammfunktion_allg = sp.integrate(f_x, x)
print(f"Allgemeine Stammfunktion: F(x) = {stammfunktion_allg} + C")

# 2. Gleichung aufstellen: F(x_wert) + C = F_wert
gleichung = sp.Eq(stammfunktion_allg.subs(x, x_wert) + C, F_wert)
print(f"Gleichung zum Lösen: {stammfunktion_allg.subs(x, x_wert)} + C = {F_wert}")

# 3. Nach C auflösen
C_loesung = sp.solve(gleichung, C)[0]
print(f"Berechnetes C: {C_loesung}")
print(f"Spezielle Stammfunktion: F(x) = {stammfunktion_allg} + ({C_loesung})\n")


# ==========================================
# 2. HIER ANPASSEN FÜR ROTATIONSKÖRPER:
f_rot = 4 - x**2
grenze_unten = -2
grenze_oben = 2
# ==========================================

print("--- ROTATIONSKÖRPER (Volumen) ---")
# V = pi * integral( f(x)^2 )
volumen_integral = sp.integrate(f_rot**2, (x, grenze_unten, grenze_oben))
volumen = sp.pi * volumen_integral
print(f"Integral von f(x)^2: {volumen_integral}")
print(f"Exaktes Volumen: {volumen} (~ {volumen.evalf():.3f})")
```
