(Nullstellen, Extrema mit Begründung & Wendepunkte)

```python
import sympy as sp

x = sp.symbols("x")
f = x**3 - 3 * x**2 + 1

f1, f2 = sp.diff(f, x), sp.diff(f, x, 2)
print(f"f'(x) = {f1}  |  f''(x) = {f2}\n")

kandidaten = sp.solve(f1, x)
print(f"Notwendige Bedingung f'(x) = 0 liefert x: {kandidaten}\n")

for k in kandidaten:
    wert_f2 = f2.subs(x, k)
    if wert_f2 > 0:
        print(f"f''({k}) = {wert_f2} > 0 -> TIEFPUNKT bei TP({k} | {f.subs(x, k)})")
    elif wert_f2 < 0:
        print(f"f''({k}) = {wert_f2} < 0 -> HOCHPUNKT bei HP({k} | {f.subs(x, k)})")

print(f"\nWendestellen f''(x) = 0: {sp.solve(f2, x)}")
```
