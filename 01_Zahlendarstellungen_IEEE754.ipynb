```python
import struct

dezimal_zahl = 42
komma_zahl = -1.17549435e-38
bits = 8

print(f"--- BASIS-UMRECHNUNGEN FÜR {dezimal_zahl} ---")
print(
    f"Binär: {bin(dezimal_zahl)} | Oktal: {oct(dezimal_zahl)} | Hex: {hex(dezimal_zahl)}"
)

print(f"\n--- ZWEIERKOMPLEMENT ---")
neg_zahl = -dezimal_zahl
zweier_komp = bin(neg_zahl & (2**bits - 1))
print(f"{neg_zahl} im {bits}-Bit Zweierkomplement: {zweier_komp}")

print(f"\n--- IEEE 754 SINGLE PRECISION (32-Bit) ---")
packed = struct.pack(">f", komma_zahl)
ieee_bits = "".join(f"{c:08b}" for c in packed)
print(f"Zahl: {komma_zahl}")
print(
    f"Vorzeichen: {ieee_bits[0]} | Exponent: {ieee_bits[1:9]} (Dez: {int(ieee_bits[1:9], 2)}) | Mantisse: {ieee_bits[9:]}"
)
```
