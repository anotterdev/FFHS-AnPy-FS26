```python
import sympy as sp

n = sp.symbols('n')
funktion = (4*n**3 + 2*n**2-1) / (5*n**3+7*n)
x0 = sp.oo  # sp.oo für Unendlich

print(f"Gesucht: lim (x -> {x0}) von f(x) = {funktion}\n")
zaehler, nenner = sp.fraction(sp.cancel(funktion))

if nenner != 1:
    lim_z = sp.limit(zaehler, x, x0)
    lim_n = sp.limit(nenner, x, x0)
    print(f"Limes Zähler: {lim_z} | Limes Nenner: {lim_n}")
    
    if (lim_z == 0 and lim_n == 0) or (lim_z in [sp.oo, -sp.oo] and lim_n in [sp.oo, -sp.oo]):
        print(f"-> Fall {lim_z}/{lim_n}! L'Hôpital anwenden:")
        print(f"f'(Zähler) = {sp.diff(zaehler, x)}")
        print(f"f'(Nenner) = {sp.diff(nenner, x)}\n")

print(f"ERGEBNIS: Grenzwert = {sp.limit(funktion, x, x0)}")
```

```python
import sympy as sp

x = sp.symbols('x')
f = sp.cos(x)
x0 = 0    # Entwicklungsstelle
grad = 3  # Grad des Polynoms

taylor = sp.series(f, x, x0=x0, n=grad).removeO()
print(f"Taylor-Polynom {grad-1}. Grades um x0={x0}:")
print(taylor)
```

```python
import sympy as sp

# ==========================================
# HIER ANPASSEN:
x = sp.symbols('x')
f = 4-x**2
# ==========================================

print("--- LÖSUNGSWEG: FLÄCHE ZWISCHEN FUNKTION UND X-ACHSE ---")
print(f"Funktion f(x) = {sp.expand(f)}\n")

print("Schritt 1: Nullstellen berechnen (Integrationsgrenzen)")
nullstellen = sp.solve(f, x)
nullstellen.sort() # Sortieren von klein nach gross
print(f"Gefundene Nullstellen bei x = {nullstellen}\n")

print("Schritt 2: Teilintegrale bilden und Beträge addieren")
gesamt_flaeche = 0

for i in range(len(nullstellen) - 1):
    unten = nullstellen[i]
    oben = nullstellen[i+1]
    
    teil_integral = sp.integrate(f, (x, unten, oben))
    teil_flaeche = sp.Abs(teil_integral) # Wichtig: Betrag nehmen!
    
    print(f"Integral von {unten} bis {oben}: {teil_integral} -> Fläche: {teil_flaeche}")
    gesamt_flaeche += teil_flaeche

print(f"\nERGEBNIS: Die absolute Gesamtfläche beträgt {gesamt_flaeche}")
```
