Bezieht sich auf Frage 1 ($q_n = q_{n-1} + 2n - 1$)

```python
import sympy as sp

# ==========================================
# HIER ANPASSEN:
n = sp.symbols("n", integer=True)
q = sp.Function("q")

# Die Rekursionsgleichung immer so umstellen, dass sie = 0 ergibt!
# Aus q(n) = q(n-1) + 2n - 1 wird: q(n) - q(n-1) - 2*n + 1
rekursion = q(n) - q(n - 1) - 2 * n + 1

# Anfangswert q(0) = 0
startwerte = {q(0): 0}
# ==========================================

print("--- LÖSUNGSWEG: EXPLIZITE FOLGE ---")
print(f"Gegebene Rekursion (gleich 0 gesetzt): {rekursion} = 0")
print(f"Mit den Startwerten: {startwerte}")

explizit = sp.rsolve(rekursion, q(n), startwerte)

print(f"\nERGEBNIS: Die direkte Formel lautet:")
print(f"q_n = {sp.expand(explizit)}")

# Teste zur Sicherheit die ersten 3 Werte
print("\nProbe:")
for i in range(3):
    print(f"n={i} -> q_{i} = {explizit.subs(n, i)}")
```
