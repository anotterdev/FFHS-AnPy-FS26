Bezieht sich auf das Lernziel "Integralrechnung: Anwendungen: Numerische Integration".

```python
import numpy as np
import scipy.integrate as spi


# ==========================================
# HIER ANPASSEN (Numerische Integration braucht Numpy statt Sympy!):
# Beispiel: Die berühmte Gausssche Glockenkurve e^(-x^2), die analytisch nicht lösbar ist.
def f_num(x):
    return np.exp(-(x**2))


grenze_unten = 0
grenze_oben = 1
# ==========================================

print("--- NUMERISCHE INTEGRATION (SciPy) ---")
print(f"Integriere von {grenze_unten} bis {grenze_oben}.")

# quad gibt ein Tupel zurück: (Berechneter Wert, Geschätzter Fehler)
integral_wert, fehler = spi.quad(f_num, grenze_unten, grenze_oben)

print(f"\nFlächeninhalt (numerisch berechnet): {integral_wert:.6f}")
print(f"Geschätzter Rechenfehler des Algorithmus: {fehler:.2e}")
print("\nBegründung:")
print(
    "Da diese Funktion schwer oder gar nicht analytisch integrierbar ist, wurde das bestimmte Integral mit einem numerischen Quadratur-Verfahren (scipy.integrate.quad) angenähert."
)
```
