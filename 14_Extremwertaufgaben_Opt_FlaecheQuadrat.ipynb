```python
import sympy as sp

# ==========================================
# HIER ANPASSEN:
x = sp.symbols("x", positive=True)  # positive=True, da eine Breite nicht negativ ist
f_parabel = 4 - x**2

# Zielfunktion aufstellen (z.B. Fläche Rechteck: Breite * Höhe)
# Breite = 2*x (von -x bis x), Höhe = f(x)
zielfunktion = 2 * x * f_parabel
# ==========================================

print("--- LÖSUNGSWEG: EXTREMWERTAUFGABE ---")
print(f"Zielfunktion A(x) = {sp.expand(zielfunktion)}\n")

A_strich = sp.diff(zielfunktion, x)
print("Schritt 1: Ableitung bilden")
print(f"A'(x) = {A_strich}\n")

print("Schritt 2: A'(x) = 0 setzen und x berechnen")
x_opt = sp.solve(A_strich, x)[0]  # Nimm den ersten positiven Wert
print(f"Optimales x (halbe Breite) = {x_opt}")

breite = 2 * x_opt
hoehe = f_parabel.subs(x, x_opt)
max_flaeche = zielfunktion.subs(x, x_opt)

print("\nSchritt 3: Abmessungen berechnen")
print(f"Breite des Rechtecks (2x): {breite}")
print(f"Höhe des Rechtecks f(x): {hoehe}")
print(f"Maximaler Flächeninhalt: {max_flaeche}")
```
