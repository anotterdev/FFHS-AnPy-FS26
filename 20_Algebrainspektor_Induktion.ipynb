```python
import sympy as sp

n = sp.symbols("n", integer=True, positive=True)

# ==========================================
# HIER ZU BEWEISENDEN TERM ANPASSEN (z.B. die linke Seite der Gleichung/Ungleichung):
term = n**2 - 2 * n - 1
start_n = 3
# ==========================================

print("--- INDUKTION: BASIS-CHECK ---")
basis_wert = term.subs(n, start_n)
print(f"Wert für n={start_n}: {basis_wert} (Sollte die Bedingung erfüllen, z.B. > 0)")

print("\n--- INDUKTION: SCHRITT (n+1) ---")
term_n1 = term.subs(n, n + 1)
print(f"Term mit (n+1) unvereinfacht: {term_n1}")

term_n1_vereinfacht = sp.expand(term_n1)
print(f"Term mit (n+1) ausmultipliziert: {term_n1_vereinfacht}")

# Differenz zwischen n+1 und n (zeigt oft das Wachstum, nützlich für Ungleichungen)
differenz = sp.simplify(term_n1 - term)
print(f"Differenz zwischen (n+1) und n: {differenz}")
```
