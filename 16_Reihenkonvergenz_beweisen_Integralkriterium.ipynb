_Bezieht sich auf Frage 7 ($\sum \frac{\ln(k)}{k}$)_

Python kann Reihen berechnen, aber der Dozent will eine **Begründung**. Oft ist das Integralkriterium der Lebensretter: Wenn das Integral von 1 bis $\infty$ konvergiert, konvergiert auch die Reihe. Divergiert das Integral, divergiert die Reihe.

Bei Frage 6 (Der Beweis, dass $1+2+4+... = -1$ ist): Dieser Beweis nutzt die Formel für die geometrische Reihe $\frac{1}{1-q}$. Diese Formel darf man aber **nur anwenden, wenn $|q| < 1$ ist.** Da hier $q = 2$ ist, ist die Anwendung der Formel illegal. Das ist ein reiner Theorie-Fakt, den du im Kopf haben musst.


```python
import sympy as sp

# ==========================================
# HIER ANPASSEN:
k = sp.symbols("k", positive=True)
folge = sp.ln(k) / k
# ==========================================

print("--- LÖSUNGSWEG: REIHENKONVERGENZ (INTEGRALKRITERIUM) ---")
print(f"Wir untersuchen die Reihe mit dem Element a_k = {folge}")
print(
    "Wir ersetzen k durch x und bilden das uneigentliche Integral von 1 bis unendlich.\n"
)

x = sp.symbols("x", positive=True)
funktion = folge.subs(k, x)

print("Schritt 1: Unbestimmtes Integral (Stammfunktion) berechnen")
stammfunktion = sp.integrate(funktion, x)
print(f"F(x) = {stammfunktion}\n")

print("Schritt 2: Uneigentliches Integral auswerten (Limes x -> unendlich)")
integral_wert = sp.integrate(funktion, (x, 1, sp.oo))
print(f"Wert des Integrals: {integral_wert}\n")

print("Schritt 3: Schlussfolgerung")
if integral_wert == sp.oo or integral_wert == -sp.oo:
    print("Das Integral ist divergent (unendlich).")
    print("-> BEGRÜNDUNG: Nach dem Integralkriterium divergiert somit auch die Reihe!")
else:
    print(f"Das Integral konvergiert gegen {integral_wert}.")
    print(
        "-> BEGRÜNDUNG: Nach dem Integralkriterium ist somit auch die Reihe konvergent!"
    )
```
