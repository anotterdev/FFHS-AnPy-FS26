```python
import sympy as sp

x = sp.symbols('x')
f_oben = -x**2 + 2*x + 1
f_unten = -x**2 + x + 1  

schnittpunkte = sp.solve(f_oben - f_unten, x)
print(f"Schnittpunkte der Kurven: x = {schnittpunkte}")

if len(schnittpunkte) >= 2:
    flaeche = sp.integrate(f_oben - f_unten, (x, schnittpunkte[0], schnittpunkte[1]))
    print(f"Flächeninhalt zwischen den Kurven: {flaeche} (~ {flaeche.evalf():.3f})")
```
