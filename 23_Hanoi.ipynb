Bei diesem Problem geht es darum, Scheiben von einem Stab auf einen anderen zu bewegen, ohne jemals eine grössere Scheibe auf eine kleinere zu legen.  
### Die Theorie:
* Rekursionsgleichung: Um $n+1$ Scheiben zu verschieben, verschiebt man erst $n$ Scheiben, dann die grösste, dann wieder die $n$ Scheiben. Daraus folgt: $a_{n+1} = 2 \cdot a_n + 1$ (mit $a_0 = 0$).  
* Explizite Formel: Die Anzahl der benötigten Züge für $n$ Scheiben ist immer exakt $a_n = 2^n - 1$.  
* Wenn nach einem Induktionsbeweis dafür gefragt wird, nutzt du dein Induktions-Skript mit behauptung = 2n - 1 und summand bzw. Rekursionsschritt $2 \cdot a_n + 1$.  

```python
# ==========================================
# HIER ANPASSEN:
anzahl_scheiben = 3
# ==========================================

print(f"--- TÜRME VON HANOI ({anzahl_scheiben} Scheiben) ---")

# Mathematische Berechnung der minimalen Züge
zuege_math = 2**anzahl_scheiben - 1
print(f"Theoretisch benötigte Züge (Formel 2^n - 1): {zuege_math}\n")

print("Ablauf der Züge:")
zug_zaehler = 0

def hanoi(n, start, ziel, zwischen):
    global zug_zaehler
    if n == 0:
        return
    
    # 1. Verschiebe n-1 Scheiben vom Start zum Zwischenstab
    hanoi(n-1, start, zwischen, ziel)
    
    # 2. Verschiebe die n-te (größte) Scheibe zum Ziel
    zug_zaehler += 1
    print(f"Zug {zug_zaehler}: Bewege Scheibe {n} von {start} nach {ziel}")
    
    # 3. Verschiebe die n-1 Scheiben vom Zwischenstab zum Ziel
    hanoi(n-1, zwischen, ziel, start)

# Funktion ausführen
hanoi(anzahl_scheiben, "Stab A", "Stab B", "Stab C")
print(f"\nTatsächlich durchgeführte Züge: {zug_zaehler}")
```
