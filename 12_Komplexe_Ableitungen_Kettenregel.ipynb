```python
import sympy as sp

# ==========================================
# HIER ANPASSEN:
x, a = sp.symbols('x a') # Alle Buchstaben, die vorkommen, deklarieren!

# 1. Spezifische Funktion ableiten (z.B. aus Kurztest 2)
funktion = sp.cos(sp.sin(sp.sqrt(x * a)))
# funktion = x**x  # Anderes Beispiel

# 2. Produktregel analysieren (g * h mehrmals ableite)
g = sp.Function('g')(x)
h = sp.Function('h')(x)
produkt = g * h
ableitungs_grad = 3
# ==========================================

print("--- 1. KOMPLEXE FUNKTION ABLEITEN ---")
print(f"Gegeben: f(x) = {funktion}")
ableitung = sp.diff(funktion, x)
print(f"f'(x) (unvereinfacht): {ableitung}")
print(f"f'(x) (vereinfacht):   {sp.simplify(ableitung)}\n")


print(f"--- 2. PRODUKTREGEL-MUSTER ({ableitungs_grad}. Ableitung) ---")
print(f"Wir leiten f(x) = g(x)*h(x) {ableitungs_grad}-mal ab:")
produkt_ableitung = sp.diff(produkt, x, ableitungs_grad)

# sp.expand klappt das aus, damit du die Binomialkoeffizienten siehst
print(sp.expand(produkt_ableitung))
```
