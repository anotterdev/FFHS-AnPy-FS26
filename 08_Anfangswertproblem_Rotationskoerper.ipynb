(Stammfunktion mit C berechnen & Rotationsvolumen)

```python
import sympy as sp

x, C = sp.symbols("x C")

# --- 1. STAMMFUNKTION MIT ANFANGSWERT ---
f_x = 3 * x**2 - 4 * x + 1
x_wert, F_wert = 1, 2  # F(1) = 2

stamm_allg = sp.integrate(f_x, x)
C_loesung = sp.solve(sp.Eq(stamm_allg.subs(x, x_wert) + C, F_wert), C)[0]
print(f"Spezielle Stammfunktion: F(x) = {stamm_allg} + ({C_loesung})\n")

# --- 2. ROTATIONSKÖRPER ---
f_rot = 4 - x**2
volumen = sp.pi * sp.integrate(f_rot**2, (x, -2, 2))
print(f"Volumen des Rotationskörpers: {volumen} (~ {volumen.evalf():.3f})")
```
