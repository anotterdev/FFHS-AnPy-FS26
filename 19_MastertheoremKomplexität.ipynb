Mastertheorem

```python
import sympy as sp

# ==========================================
# HIER ANPASSEN: T(n) <= d * T(n/b) + c*n
d = 3  # Anzahl der Teilprobleme (z.B. 3 bei Karatsuba)
b = 2  # Faktor der Verkleinerung (z.B. 2 bei Halbierung)
# ==========================================

print("--- MASTERTHEOREM CHECKER ---")
if d < b:
    print(f"Fall 1 (d < b): O(n)")
elif d == b:
    print(f"Fall 2 (d == b): O(n * log(n))")
else:
    # d > b
    exponent = sp.log(d, b)
    print(f"Fall 3 (d > b): O(n^(log_{b}({d})))")
    print(f"Exakter Exponent: {exponent}")
    print(f"Dezimal: O(n^{exponent.evalf():.3f})")
```
