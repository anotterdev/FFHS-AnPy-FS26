```python
import sympy as sp


# --- 1. REKURSION: FAKULTÄT & FIBONACCI ---
def factorial_rekursiv(n):
    return 1 if n == 1 else n * factorial_rekursiv(n - 1)


def fib_rekursiv(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib_rekursiv(n - 1) + fib_rekursiv(n - 2)


# --- 2. INDUKTIONS-ASSISTENT ---
n = sp.symbols("n", integer=True, positive=True)
term = n**2 - 2 * n - 1  # Zu beweisender Term
start_n = 3

print("\n--- INDUKTION: BASIS-CHECK ---")
print(f"Wert für n={start_n}: {term.subs(n, start_n)}")

print("\n--- INDUKTION: SCHRITT (n+1) ---")
term_n1 = term.subs(n, n + 1)
print(f"Term mit (n+1) unvereinfacht: {term_n1}")
print(f"Term mit (n+1) ausmultipliziert: {sp.expand(term_n1)}")
print(f"Differenz zwischen (n+1) und n: {sp.simplify(term_n1 - term)}")
```
