Landau-Symbole ($\mathcal{O}$-Notation)

Bei diesen Aufgaben geht es darum, wie schnell Algorithmen bei riesigen Datenmengen wachsen. Konstanten und kleinere Terme werden rigoros ignoriert.

**Die absolute Wachstums-Hierarchie (von schnell nach extrem langsam):**

- $\mathcal{O}(1)$ (Konstant)
    
- $\mathcal{O}(\log(n))$ (Logarithmisch)
    
- $\mathcal{O}(n)$ (Linear)
    
- $\mathcal{O}(n \log(n))$ (Linear-logarithmisch, z.B. Merge-Sort)
    
- $\mathcal{O}(n^2)$ (Quadratisch, z.B. Bubble-Sort)
    
- $\mathcal{O}(n^3)$ (Kubisch)
    
- $\mathcal{O}(2^n)$ (Exponentiell)
    
- $\mathcal{O}(3^n)$ (Exponentiell mit grösserer Basis)
    
- $\mathcal{O}(n!)$ (Fakultät)
    
- $\mathcal{O}(n^n)$ (Am schlimmsten)
    

**Wichtige Regeln für Gleichheit ($=$) und Teilmengen ($\subsetneq$):**

- **Faktoren sind egal:** $\mathcal{O}(n) = \mathcal{O}(5n) = \mathcal{O}(n + 1000)$
    
- **Logarithmus-Basen sind egal:** $\mathcal{O}(\log_2(n)) = \mathcal{O}(\log_{10}(n)) = \mathcal{O}(\ln(n))$
    
- **Niedrigere Potenzen fallen weg:** $\mathcal{O}(n^2 + n) = \mathcal{O}(n^2)$
    
- **Exponentielle Basen sind NICHT egal:** $\mathcal{O}(2^n) \subsetneq \mathcal{O}(3^n)$
    
- **Strenge Teilmenge (echt kleiner):** $\mathcal{O}(n^2) \subsetneq \mathcal{O}(n^3)$ bedeutet, dass $n^2$ strikt langsamer wächst als $n^3$.
