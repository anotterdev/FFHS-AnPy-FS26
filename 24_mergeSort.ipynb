# Merge Sort 
* teilt eine Liste immer wieder in zwei Hälften, bis nur noch Einzelelemente da sind, und "mischt" (merged) diese dann sortiert wieder zusammen.  
### Die Theorie:
* Rekursionsgleichung: Die Liste wird in 2 Teile geteilt (Faktor $b=2$), es entstehen 2 Teilprobleme (Faktor $d=2$), und das Mischen kostet linearen Aufwand $n$. 
* Die Formel lautet: $T(n) \le 2 \cdot T(n/2) + \mathcal{O}(n)$.  
* Komplexität (Mastertheorem): Da $d = b$ ($2 = 2$), greift Fall 2 des Mastertheorems. Die Laufzeit ist immer $\mathcal{O}(n \log_2(n))$.  
* Merge Sort vs. Bubble Sort: Bubble Sort hat $\mathcal{O}(n^2)$. Merge Sort ist mit $\mathcal{O}(n \log(n))$ bei grossen Datenmengen immer massiv schneller.  

```python
# ==========================================
# HIER ANPASSEN:
meine_liste = [38, 27, 43, 3, 9, 82, 10]
# ==========================================

print("--- MERGE SORT ALGORITHMUS ---")
print(f"Unsortierte Originalliste: {meine_liste}\n")


def merge_sort(liste):
    # Abbruchbedingung: Liste mit 1 oder 0 Elementen ist bereits sortiert
    if len(liste) <= 1:
        return liste

    # Mitte finden und aufteilen
    mitte = len(liste) // 2
    links = merge_sort(liste[:mitte])
    rechts = merge_sort(liste[mitte:])

    # Mischen (Merge)
    return merge(links, rechts)


def merge(links, rechts):
    sortiert = []
    i = j = 0

    # Elemente vergleichen und das kleinere in die neue Liste packen
    while i < len(links) and j < len(rechts):
        if links[i] <= rechts[j]:
            sortiert.append(links[i])
            i += 1
        else:
            sortiert.append(rechts[j])
            j += 1

    # Restliche Elemente anhängen (falls eine Liste schon leer ist)
    sortiert.extend(links[i:])
    sortiert.extend(rechts[j:])

    return sortiert


# Funktion ausführen
sortierte_liste = merge_sort(meine_liste)
print(f"Sortierte Liste: {sortierte_liste}")
```
