ddf

```python
import scipy.optimize as opt


def f(x):
    return x**3 - 2 * x - 5

def f_ableitung(x):
    return 3 * x**2 - 2

startwert = 2.0
try:
    nullstelle = opt.newton(f, x0=startwert, fprime=f_ableitung)
    print(f"Gefundene Nullstelle (Newton): x = {nullstelle}")
except RuntimeError:
    print("Verfahren konvergiert nicht.")
```
