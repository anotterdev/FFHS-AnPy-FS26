```python
import sympy as sp

# ==========================================
# HIER ANPASSEN:
x = sp.symbols('x')
# Beispiel: f(x) = |x| / x (Klassischer Fall für eine Sprungstelle)
funktion = sp.Abs(x) / x  
x0 = 0  # Die kritische Stelle, die auf Stetigkeit geprüft wird
# ==========================================

print(f"--- LÖSUNGSWEG: STETIGKEITSPRÜFUNG AN DER STELLE x0 = {x0} ---")

# 1. Linksseitiger Grenzwert (x nähert sich von unten)
lim_links = sp.limit(funktion, x, x0, dir='-')
print(f"Linksseitiger Grenzwert (x -> {x0}-): {lim_links}")

# 2. Rechtsseitiger Grenzwert (x nähert sich von oben)
lim_rechts = sp.limit(funktion, x, x0, dir='+')
print(f"Rechtsseitiger Grenzwert (x -> {x0}+): {lim_rechts}")

# 3. Stetigkeit bewerten
print("\nFAZIT:")
if lim_links == lim_rechts:
    print(f"Da der links- und rechtsseitige Grenzwert übereinstimmen ({lim_links}), ist die Funktion an der Stelle x0={x0} stetig fortsetzbar.")
else:
    print(f"Da {lim_links} != {lim_rechts}, liegt an der Stelle x0={x0} eine UNSTETIGKEIT (Sprungstelle) vor!")
```
